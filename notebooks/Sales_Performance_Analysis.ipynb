# SALES PERFORMANCE ANALYSIS

### Importing the necessary python libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Reading the Dataset

```python
df=pd.read_excel("C://Users//akisp//OneDrive//Documentos//super store sales//Orders.xlsx")
df
```

### EDA (Exploratory Data Analysis)

```python
df.head()
```

#### Giving correct dtypes

```python
df.info()
```

#### Removing Duplicates

```python
df=df.drop_duplicates()
```

#### Filling Missing Values

```python
df.isnull().sum()
print(df.isnull().sum())
df=df.dropna()
```

#### Fix Date Format

```python
df['Order Date']=pd.to_datetime(df['Order Date'])
```

#### Creating new Columns

```python
#Create new columns
df['Year']=df['Order Date'].dt.year
df['Month']=df['Order Date'].dt.month
df['Month_Name']=df['Order Date'].dt.month_name()
df['Profit Margin']=df['Profit']/df['Sales']
```

#### KPI Calculations

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
avg_order_value = df['Sales'].mean()
profit_margin = total_profit / total_sales

print("Total Sales:", round(total_sales,2))
print("Total Profit:", round(total_profit,2))
print("Average Order Value:", round(avg_order_value,2))
print("Profit Margin:", round(profit_margin,2))
```

### Data Visulaizations/Trend Analysis

#### Monthly Sales Trend

```python
monthly_sales = df.groupby(['Year','Month'])['Sales'].sum().reset_index()

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.plot(monthly_sales['Month'], monthly_sales['Sales'])
#plt.savefig("monthly_sales_trend.png", dpi=300, bbox_inches='tight')
plt.savefig("chart.png")
plt.show()
```

#### Region-Wise Sales

```python

region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

plt.figure()
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
region_sales.plot(kind='bar',rot=0)
plt.savefig("region_sales_trend.png", dpi=300, bbox_inches='tight')
plt.show()
```

#### Top 5 Products

```python
#Top Products
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(5)

plt.figure()
plt.title("Top 5 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")
top_products.plot(kind='barh')
plt.savefig("Top_Product_Analysis.png", dpi=300, bbox_inches='tight')
plt.show()
```

#### Profitabaility Analysis

```python
#Profitability Analysis
category_profit = df.groupby('Category')['Profit'].sum()

plt.figure()
plt.title("Profit by Category")
plt.xlabel("Category")
#plt.xticks(rotation=30,ha='right')
plt.ylabel("Profit")
category_profit.plot(kind='bar',rot=0)
plt.savefig("category_profit_Analysis.png", dpi=300, bbox_inches='tight')
plt.show()
```

#### Business Insights

```python
print("INSIGHTS:")
print("1. Highest sales region:", region_sales.idxmax())
print("2. Most Profitable Category:", category_profit.idxmax())
print("3. Top selling product:", top_products.idxmax())
```
